```python
# if faststtext embeddings are used 
from gensim.models.fasttext import load_facebook_vectors
model = load_facebook_vectors("/Users/carmand/Desktop/RS/enp_embeddings/toastynews.bin")

query_terms = [
    "revolution",
    "student",
    "newspaper",
]

topn = 20
rows = []

for term in query_terms:
    if term not in model:
        rows.append({
            "query_term": term,
            "similar_term": None,
            "similarity": None,
            "note": "term not in vocabulary"
        })
        continue

    for similar_term, score in model.most_similar(term, topn=topn):
        rows.append({
            "query_term": term,
            "similar_term": similar_term,
            "similarity": score,
            "note": ""
        })

df = pd.DataFrame(rows)
df.to_csv("expanded_queries.csv", index=False, encoding="utf-8-sig")

print(df)
```

```python
# multiword query example

phrase = ["returned student"]

valid = [w for w in phrase if w in model]

if valid:
    results = model.most_similar(positive=valid, topn=20)
    for word, score in results:
        print(word, score)
```
