```python
# load the pipeline

from impresso_pipelines.ocrqa import OCRQAPipeline
```

```python
# Initialize the pipeline

ocrqa_pipeline = OCRQAPipeline()
```

```python
# test the pipeline with a sample text 

text = "Mr 5. Zee former secretary of the local Chinese has been appointed general secre-                 Itary oi the Chinese General Chamber of Commerce Mr Zee is on American returned student"
result = ocrqa_pipeline(text, diagnostics=True)
print(result)
```

```python
import pandas as pd

# Load CSV
df = pd.read_csv("/data/proquest.csv")

# Keep only the first 5 rows for testing
df_test = df.head(5).copy()

# Initialize OCRQA pipeline
# ocrqa_pipeline = OCRQAPipeline()

# Compute OCR quality score for each row
df_test["ocr_score"] = df_test["fulltext"].fillna("").apply(
    lambda x: ocrqa_pipeline(x)["score"]
)

# (Optional) also keep detected language
df_test["language"] = df_test["fulltext"].fillna("").apply(
    lambda x: ocrqa_pipeline(x)["language"]
)

# Save updated CSV
df_test.to_csv("proquest_with_ocr_scores_test.csv", index=False)

print(df_test.head())
```

```python
# apply the pipe to a larger corpus

import pandas as pd
# from impresso_pipelines.ocrqa import OCRQAPipeline

# Load CSV
df = pd.read_csv("/data/proquest_exp_ft.csv")

# Initialize OCRQA pipeline
# ocrqa_pipeline = OCRQAPipeline()

# Compute OCR quality score for each row
df["ocr_score"] = df["Text"].fillna("").apply(
    lambda x: ocrqa_pipeline(x, language="en")["score"]
)

# Optional: store the fixed language value
df["language"] = "en"

# Save updated CSV
df.to_csv("proquest_with_ocr_scores.csv", index=False)

print(df.head())
```
